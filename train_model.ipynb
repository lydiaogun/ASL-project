```python
import numpy as np
import pandas as pd
from sklearn import metrics
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score
from sklearn.metrics import classification_report, confusion_matrix
import matplotlib.pyplot as plt
import joblib
from landmarks import normalise
from collections import Counter





def load_and_normalise(filename):
    df = pd.read_csv(filename)

    nrows= []
    nlabels = []
    fails = 0


    for index, row in df.iterrows():
        values = row.iloc[0:63].tolist()
        handedness= row.iloc[-2]
        try:
            result = normalise(values, handedness, scale= False)
            nrows.append(result)
            nlabels.append((row.iloc[-1]).upper())

        except ValueError:
            fails +=1

    return nrows, nlabels, fails

rows, labels, fails = load_and_normalise('dataset_kaggle.csv')

knn = KNeighborsClassifier(n_neighbors=5)
knn.fit(rows, labels)
test_rows, test_labels, test_fails = load_and_normalise('dataset.csv')
y_pred = knn.predict(test_rows) #takes 484 normalised hands and returns 484 letter guesses.
accuracy = accuracy_score(test_labels, y_pred) #lines up the true letters against the guesses and counts how many match, as a fraction.
print(accuracy)
confusion_mat = metrics.confusion_matrix(test_labels, y_pred)
labels_sorted = sorted(set(test_labels))
pd.set_option('display.max_columns', None)
pd.set_option('display.width', 200)
print("Confusion Matrix:")
print(pd.DataFrame(confusion_mat, index=labels_sorted, columns=labels_sorted))
class_report = classification_report(test_labels, y_pred)
print("Classification Report:")
print(class_report)
print(Counter(labels))
```
